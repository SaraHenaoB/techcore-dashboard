```python
import pandas as pd
import numpy as np
import openpyxl
from pathlib import Path
```

```python
#Extraer el dataset original verificación de las primeras 5 filas
PROJECT_ROOT = Path.cwd().parent
file_path = PROJECT_ROOT / "data" / "processed" / "ventasTransformed.csv"
#asegurar el tipo de dato float para columnas decimales
column_type={"PrecioUnitarioProducto1": float, "SubtotalProducto1": float, "PrecioUnitarioProducto2": float,
    "SubtotalProducto2": float,"PrecioUnitarioProducto3": float, "SubtotalProducto3": float, "TotalVenta": float}
dataset = pd.read_csv(file_path, sep= "\t", dtype= column_type, encoding="utf-8")
#Leer la hora con formato  AM/PM
dataset["HoraVenta"]=pd.to_datetime(dataset["HoraVenta"], format="%I:%M:%S %p").dt.time
dataset.head()
```

```python
#verificación 5 últimas filas
dataset.tail()
```

```python
#Verificación general de toda la data
dataset.info()
```

Creación de tablas: Ciudad, Vendedores, Clientes, Productos, Fechas, Sucursales, Facturas y Detalles Facturas.

```python
#Tabla ciudad

cities= sorted(dataset["CiudadSucursal"].unique())
cityID = range(1, len(cities)+1)
dt_City= pd.DataFrame({"CiudadID": cityID ,"NombreCiudad": cities})
dt_City.head()

```

```python
#Tabla vendedores
salesperson = sorted(dataset["VendedorNombre"].unique())
SPersonID = range(1, len(salesperson)+1)
dt_SalesPerson = pd.DataFrame({"VendedorID": SPersonID, "NombreVendedor": salesperson})
dt_SalesPerson.head()
```

```python
#Tabla Clientes
dt_Client = dataset[["ClienteNombre","GeneroCliente", "EdadCliente", "TelefonoCliente", "EmailCliente", "DireccionCliente"]]
dt_Client = dt_Client.drop_duplicates(subset="ClienteNombre").reset_index(drop=True)
dt_Client["ClienteID"] = range(1, len(dt_Client["ClienteNombre"])+1)
columnOrder = ["ClienteID","ClienteNombre","GeneroCliente", "EdadCliente", "TelefonoCliente", "EmailCliente", "DireccionCliente"]
dt_Client = dt_Client[columnOrder]


dt_Client.head()
```

```python
#Tabla Productos
productslist = pd.concat([dataset["NombreProducto1"], dataset["NombreProducto2"], dataset["NombreProducto3"]], ignore_index=True)
brandlist = pd.concat([dataset["MarcaProducto1"], dataset["MarcaProducto2"], dataset["MarcaProducto3"]], ignore_index=True)
pricelist = pd.concat([dataset["PrecioUnitarioProducto1"], dataset["PrecioUnitarioProducto2"], dataset["PrecioUnitarioProducto3"]], ignore_index=True)

dt_temporary = pd.DataFrame({"NombreProducto": productslist, "MarcaProducto": brandlist, "PrecioUnitario": pricelist})
dt_product = dt_temporary.drop_duplicates(subset="NombreProducto").reset_index(drop=True)
dt_product["ProductoID"] = range(1, len(dt_product["NombreProducto"])+1)
columnOrderP = ["ProductoID", "NombreProducto", "MarcaProducto", "PrecioUnitario"]
dt_product = dt_product[columnOrderP]
dt_product.head()

```

```python
#Tabla fecha
dataset["FechaVenta"] = pd.to_datetime(dataset["FechaVenta"])
startYear= dataset["FechaVenta"].min().year
endYear = dataset["FechaVenta"].max().year
print(f"Primer año de venta:{startYear}. Último año de venta registrado: {endYear}")

completeRange = pd.date_range(start=f"{startYear}-01-01", end=f"{endYear}-12-31")
dt_temporary_date = pd.DataFrame({"FechaBase": completeRange})

monthName = {1:"Enero", 2:"Febrero", 3:"Marzo", 4:"Abril", 5:"Mayo", 6:"Junio", 7:"Julio",
          8:"Agosto", 9:"Septiembre", 10:"Octubre", 11:"Noviembre", 12:"Diciembre"}

dayName = {0:"Lunes", 1:"Martes", 2:"Miercoles", 3:"Jueves", 4:"Viernes", 5:"Sabado", 6: "Domingo"}

dt_date = pd.DataFrame({"FechaID": dt_temporary_date["FechaBase"].dt.strftime('%Y%m%d').astype(int),
                        "FechaVenta": dt_temporary_date["FechaBase"].dt.strftime('%Y/%m/%d') ,
                        "AñoVenta": dt_temporary_date["FechaBase"].dt.year ,
                        "MesVenta": dt_temporary_date["FechaBase"].dt.month ,
                        "MesNombre": dt_temporary_date["FechaBase"].dt.month.map(monthName) ,
                        "DiaVenta": dt_temporary_date["FechaBase"].dt.day,
                        "DiaNombre": dt_temporary_date["FechaBase"].dt.dayofweek.map(dayName),
                        "Trimeste": 'Q'+dt_temporary_date["FechaBase"].dt.quarter.astype(str),
                        "EsFinDeSemana": dt_temporary_date["FechaBase"].dt.dayofweek.isin([5,6]).map({True: "Si",False: "No"})})

dt_date.head()
```

```python
dt_date.tail()
```

```python
#Tabla Sucursal
dt_branch = dataset[["SucursalNombre", "CiudadSucursal"]].drop_duplicates(subset="SucursalNombre").reset_index(drop=True)
dt_branch["SucursalID"] = range(1, len(dt_branch)+1)
#Búsqueda del ID en la tabla de ciudad, funciona igual que un join
dt_branch = dt_branch.merge(right= dt_City, left_on="CiudadSucursal", right_on="NombreCiudad", how= "left")
columnOrderS = ["SucursalID", "CiudadID", "SucursalNombre"]
dt_branch = dt_branch[columnOrderS]

dt_branch.head()
```

```python
#Tabla Facturas
dt_invoice = dataset[["TotalVenta","HoraVenta","MetodoPago","DescuentoVenta","FechaVenta", "SucursalNombre","VendedorNombre","ClienteNombre"]]
dt_invoice["FacturaID"] = range(1, len(dt_invoice)+1)
#Llamar a todas las claves
dt_invoice = dt_invoice.merge(right=dt_Client[["ClienteID","ClienteNombre"]],left_on="ClienteNombre",right_on="ClienteNombre", how="left")
dt_invoice = dt_invoice.merge(right=dt_SalesPerson,left_on="VendedorNombre",right_on="NombreVendedor", how="left")
dt_invoice = dt_invoice.merge(right=dt_branch[["SucursalID","SucursalNombre"]], left_on="SucursalNombre", right_on="SucursalNombre", how="left")
dt_invoice["FechaVenta"] = dt_invoice["FechaVenta"].dt.strftime('%Y/%m/%d')
dt_invoice = dt_invoice.merge(right=dt_date[["FechaVenta","FechaID"]], left_on="FechaVenta",right_on="FechaVenta",how="left")
dt_invoice = dt_invoice.rename(columns={"DescuentoVenta": "Descuento"})
columnOrderI = ["FacturaID","ClienteID","VendedorID","SucursalID","FechaID", "HoraVenta", "MetodoPago", "Descuento", "TotalVenta"]
dt_invoice = dt_invoice[columnOrderI]
dt_invoice.head()
```

```python
dt_invoice.tail()
```

```python
#Tabla Detalle Factura
dataset["FilaPuente"] = range(1, len(dataset)+1)
dt_detailInvoice = pd.concat([dataset[["FilaPuente","NombreProducto1","CantidadProducto1","PrecioUnitarioProducto1","SubtotalProducto1"]]
                              .set_axis(["FilaPuente","NombreProducto","Cantidad","PrecioUnitario", "subtotal"], axis=1)
                              ,dataset[["FilaPuente","NombreProducto2","CantidadProducto2","PrecioUnitarioProducto2","SubtotalProducto2"]]
                              .set_axis(["FilaPuente","NombreProducto","Cantidad","PrecioUnitario", "subtotal"], axis=1)
                              , dataset[["FilaPuente","NombreProducto3","CantidadProducto3","PrecioUnitarioProducto3","SubtotalProducto3"]]
                              .set_axis(["FilaPuente","NombreProducto","Cantidad","PrecioUnitario", "subtotal"], axis=1)]
                              , ignore_index=True)

dt_detailInvoice = dt_detailInvoice[dt_detailInvoice["NombreProducto"]!="No registrado"].reset_index(drop = True)

dt_detailInvoice ["DetalleID"] = range(1, len(dt_detailInvoice)+1)
dt_detailInvoice = dt_detailInvoice.merge(right=dt_product[["ProductoID","NombreProducto"]], left_on="NombreProducto"
                                          , right_on="NombreProducto", how="left")
dt_detailInvoice = dt_detailInvoice.merge(right=dt_invoice["FacturaID"],left_on="FilaPuente", right_on="FacturaID", how="left")
columnOrderDI = ["DetalleID","FacturaID","ProductoID","Cantidad","PrecioUnitario", "subtotal"]
dt_detailInvoice = dt_detailInvoice[columnOrderDI]
dt_detailInvoice.head()
```

```python
dt_detailInvoice.tail()
```

```python
#Exportar a un archivo de excel
Final_File = PROJECT_ROOT / "data" / "processed" / "modeloVentas.xlsx"

# 2. Creamos el escritor de Excel de Pandas
with pd.ExcelWriter(Final_File, engine='openpyxl') as writer:
    # Guardamos las 6 Tablas  sin el índice de Python
    dt_City.to_excel(writer, sheet_name="Ciudad", index=False)
    dt_SalesPerson.to_excel(writer, sheet_name="Vendedores", index=False)
    dt_Client.to_excel(writer, sheet_name="Clientes", index=False)
    dt_product.to_excel(writer, sheet_name="Productos", index=False)
    dt_date.to_excel(writer, sheet_name="Fechas", index=False)
    dt_branch.to_excel(writer, sheet_name="Sucursales", index=False)
    
    # Guardamos las 2 Tablas de Hechos
    dt_invoice.to_excel(writer, sheet_name="Facturas", index=False)
    dt_detailInvoice.to_excel(writer, sheet_name="Detalles Facturas", index=False)

print(f"Se ha creado el archivo '{Final_File.name}' con las 8 tablas distribuidas en pestañas.")
```
